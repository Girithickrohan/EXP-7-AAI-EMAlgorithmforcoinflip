```python
from scipy.stats import binom

def em(tol=1e-8):
    exp = [(5,5),(9,1),(8,2),(4,6),(7,3)]
    a, b = .6, .5
    it = 0

    print("--- EXPERIMENT DATA ---")
    for i,(h,t) in enumerate(exp,1):
        print(f"E{i}: Heads = {h}, Tails = {t}")

    while True:
        it += 1
        ha=ta=hb=tb=0

        if it == 1:
            print("\n--- FIRST E-STEP ---")

        for i,(h,t) in enumerate(exp,1):
            n = h+t
            la, lb = binom.pmf(h,n,a), binom.pmf(h,n,b)
            pa = la/(la+lb)
            pb = 1-pa

            ha += pa*h
            ta += pa*t
            hb += pb*h
            tb += pb*t

            if it == 1:
                print(f"E{i}: P(A|E) = {pa:.4f}, P(B|E) = {pb:.4f}")

        na, nb = ha/(ha+ta), hb/(hb+tb)

        if it == 1:
            print("\n--- FIRST M-STEP ---")
            print(f"Coin A: Expected Heads = {ha:.4f}, Expected Tails = {ta:.4f}")
            print(f"Coin B: Expected Heads = {hb:.4f}, Expected Tails = {tb:.4f}")
            print(f"New Theta A = {na:.4f}")
            print(f"New Theta B = {nb:.4f}")

        if abs(na-a)<tol and abs(nb-b)<tol:
            a,b = na,nb
            break

        a,b = na,nb

    print("\n--- CONVERGENCE REACHED ---")
    print(f"Algorithm converged after {it} iterations.")
    print(f"Final Theta A = {a:.4f}")
    print(f"Final Theta B = {b:.4f}")
    print(f"Coin A probability of Heads = {a:.2f}")
    print(f"Coin B probability of Heads = {b:.2f}")

em()
```
